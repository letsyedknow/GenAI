## Example few shot and customer ticket analysis

### classification of category

```python
classification_categories = ["billing inquiry", 
                             "technical support", 
                             "account management", 
                             "Product feedback"]
```

### Prompt template

```python
User_message = "I am unable to access my account after the recent update."
prompt_template = f"""
You are an expert support ticket classifier for a software company. 
Your task is to analyze the user's message and classify it into one of the following predefined categories:
{', '.join(classification_categories)}

You must respond with only the category name and nothing else.

--
User Message: "I am having trouble logging into my account. Can you help me reset my password?
Classification: "account management"
User Message: "The software crashes every time I try to open it. What should I do?"
Classification: "technical support"
User Message: "I would like to update my billing information for my subscription."
Classification: "billing inquiry"
User Message: "I love the new features in the latest update! Great job!"
Classification: "Product feedback"

Now, classify the following message:
User Message: {{user_message}}
Classification:
"
"""

print(prompt_template)
```

```python
def classify_message(user_message):
    prompt = prompt_template.replace("{{user_message}}", user_message)
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
    )
    # predictive_text = response.text.strip()
    # if predictive_text in classification_categories:
    #     return predictive_text
    # else:
    #     print(f"Warning: Model output not in predefined categories: '{predictive_text}'")
    return response.text.strip()
```

```python
print(classify_message('I am getting charged twice for my subscription this month.'))
```

```python
test_messages = [
    "I need help updating my credit card information.",
    "The app keeps freezing whenever I try to use it.",
    "Can you tell me how to cancel my subscription?",
    "Great job on the new update! I love the new design.",
    "Why am I being charged extra this month?"
]

# Call the classify_message function for each test message
for message in test_messages:
    classification = classify_message(message)
    print(f"User Message: '{message}'\nClassification: '{classification}'\n")
```
