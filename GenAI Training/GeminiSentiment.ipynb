The problem is to create an automated system that can accurately determine the sentiment of a user's message.

The system needs to:

Take a piece of text (a user's message) as input.
Analyze the text to understand its emotional tone.
Classify the sentiment into one of three predefined categories: "positive", "negative", or "neutral".
Return only the name of the identified category as the final output.
The provided code accomplishes this by using a "few-shot" prompt to instruct the Gemini language model on how to perform this sentiment analysis task, giving it clear examples to follow.

```python
from google import genai

client = genai.Client()
```

```python
sentiment_categories = ["positive", "negative", "neutral"]

sentiment_prompt_template = f"""
You are an expert sentiment analyzer. 
Your task is to analyze the sentiment of the user's message and classify it into one of the following predefined categories:
{', '.join(sentiment_categories)}

You must respond with only the category name and nothing else.

--
User Message: "I love the new features in the latest update! Great job!"
Sentiment: "positive"
User Message: "The app keeps crashing, and I am very frustrated."
Sentiment: "negative"
User Message: "The update is okay, but I didn't notice much difference."
Sentiment: "neutral"

Now, classify the following message:
User Message: {{user_message}}
Sentiment:
"
"""

def analyze_sentiment(user_message):
    prompt = sentiment_prompt_template.replace("{{user_message}}", user_message)
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
    )
    return response.text.strip()

# Example usage
test_sentiments = [
    "My app is not responding after the update.",
    "This is the worst experience I've had.",
    "It's fine, nothing special.",
]

for message in test_sentiments:
    sentiment = analyze_sentiment(message)
    print(f"User Message: '{message}'\nSentiment: '{sentiment}'\n")
```
