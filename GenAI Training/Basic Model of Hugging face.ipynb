```python
import os
from openai import OpenAI
from config import cred


def generate_text(prompt, model="zai-org/GLM-4.5:novita", max_tokens=1000):
    """
    Generate text using the specified model.
    
    :param prompt: The input prompt for text generation.
    :param model: The model to use for generation.
    :param max_tokens: The maximum number of tokens to generate.
    :return: Generated text as a string.
    """
    client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.getenv("HF_TOKEN"),
)
    
    try:
        completion = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=max_tokens
        )
        return completion.choices[0].message
    except Exception as e:
        return f"An error occurred: {e}"
    


if __name__ == "__main__":
    user_prompt = input("Enter a prompt: ")
    result = generate_text(user_prompt)
    print("Generated Text:")
    # print(type(result))
    print(result)
```

## Do a sentiment analysis

```python

sentiment_categories = ["positive", "negative", "neutral"]

sentiment_prompt_template = f"""
You are an expert sentiment analyzer. 
Your task is to analyze the sentiment of the user's message and classify it into one of the following predefined categories:
{', '.join(sentiment_categories)}

You must respond with only the category name and nothing else.

--
User Message: "I love the new features in the latest update! Great job!"
Sentiment: "positive"
User Message: "The app keeps crashing, and I am very frustrated."
Sentiment: "negative"
User Message: "The update is okay, but I didn't notice much difference."
Sentiment: "neutral"

Now, classify the following message:
User Message: {{user_message}}
Sentiment:
"""

def analyze_sentiment(user_message):
    prompt = sentiment_prompt_template.replace("{{user_message}}", user_message)
    response = generate_text(prompt, max_tokens=100)
    return response

# Example usage
test_sentiments = [
    "I absolutely love this app!",
    "This is the worst experience I've had.",
    "It's fine, nothing special.",
]

for message in test_sentiments:
    sentiment = analyze_sentiment(message)
    print(f"User Message: '{message}'\nSentiment: '{sentiment}'\n")
```
